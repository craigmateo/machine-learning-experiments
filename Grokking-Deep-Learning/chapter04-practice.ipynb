```python
input = 0.75
weight = 0.5
target = 0.8
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    # Calculate the prediction
    prediction = input * weight

    # Calculate the error
    error = target - prediction

    # Update the weight
    weight += learning_rate * error * input

    # Print the current state
    print(f"Iteration {i+1}: Prediction={prediction:.4f}, Error={error:.4f}, Weight={weight:.4f}")

    if abs(error) < 0.01:
        print("Converged!")
        break
```
